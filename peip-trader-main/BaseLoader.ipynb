### Classe Loader

Classe pour charger les valeurs mobilières.

```python
class Loader :

    # 1.ajouter vos valeurs mobilières ici
    ADA = "Cardano"
    AAVE = "Aave"
    BTC = "Bitcoin"
    COSMOS = "Cosmos"
    DOGECOIN = "Dogecoin"
    ETH = "Ethereum"
    LITECOIN = "Litecoin"
    MONERO = "Monero"
    POLKADOT = "Polkadot"
    USDCOIN = "USDCoin"
    SOLANA = "Solana"
    
    AIRBUS = "Airbus"
    AMAZON = "Amazon"
    AMUNDI = "Amundi"
    APPLE = "Apple"
    ATOS = "Atos"
    ATT = "AT&T"
    AXA = "AXA"
    CAC40 = "cac40"
    CREDIT_AGRICOLE = "CreditAgricole"
    GAMESTOP = "Gamestop"
    INTEL = "Intel"
    MICROSOFT = "Microsoft"
    NETFLIX = "Netflix"
    NOKIA = "Nokia"
    NVIDIA = "Nvidia"
    ORANGE = "Orange"
    ORPEA = "Orpea"
    SP500 = "Sp500"
    STELLANTIS = "Stellantis"
    TESLA = "Tesla"
    TOTAL = "Total"
    VINCI = "VINCI"

    GOLD = "Gold"

    ETF_SP500_TR = "ETFSP500TR"
    ETF_ISHARE_CORE_SP500 = "ETFiSharesCoreSP500"
    ETF_ISHARE_MSCI_CORE_MSCI_WORLD = "ETFisharesCoreMSCIWorld"
    ETF_ISHARE_GROWTH_SP500 = "ETFiSharesGrowthSP500"
    ETF_ISHARE_SEMICONDUCTOR = "ETFiSharesSemiconductor"
    ETF_ISHARE_MSCI_WORLD_INDEX = "ETFiSharesMSCIWorldIndex"
    ETF_VANGUARD_GROWTH_INDEX_FUND = "ETFVanguardGrowthIndexFund"
    ETF_VANGUARD_HIGH_DIVIDEND = "ETFVanguardHighDividendYieldIndexFund"
    ETF_VANGUARD_REAL_ESTATE = "ETFVanguardRealEstateIndexFund"
    ETF_VANGUARD_SP500 = "ETFVanguardSP500"
    ETF_AMUNDI_CW8 = "ETFAmundiCW8"

    # 2. ajouter la variable dans la liste
    SECURITY = [ADA, 
                AAVE,
                AIRBUS,
                AMAZON,
                AMUNDI,
                APPLE,
                ATOS,
                ATT,
                AXA, 
                BTC, 
                CAC40,
                CREDIT_AGRICOLE,
                COSMOS, 
                DOGECOIN, 
                ETF_AMUNDI_CW8,
                ETF_ISHARE_CORE_SP500,  
                ETF_ISHARE_GROWTH_SP500, 
                ETF_ISHARE_MSCI_CORE_MSCI_WORLD,
                ETF_ISHARE_MSCI_WORLD_INDEX,
                ETF_ISHARE_SEMICONDUCTOR,
                ETF_SP500_TR,
                ETF_VANGUARD_GROWTH_INDEX_FUND,
                ETF_VANGUARD_HIGH_DIVIDEND,
                ETF_VANGUARD_REAL_ESTATE,
                ETF_VANGUARD_SP500,
                ETH, 
                GAMESTOP,
                GOLD,
                INTEL, 
                LITECOIN, 
                MONERO, 
                POLKADOT, 
                SOLANA,
                MICROSOFT, 
                NETFLIX, 
                NVIDIA, 
                ORANGE, 
                ORPEA, 
                SP500, 
                STELLANTIS,
                TESLA, 
                TOTAL,
                VINCI]

    # 3. ajouter votre source ici (dossier `nouvelleSource` dans `data`)
    COINCODEX = "Coincodex"
    GOOGLE_FINANCE = "GoogleFinance"
    YAHOO = "finance.yahoo.com"
    KAGGLE = "Kaggle"
    NASDAQ = "Nasdaq"
    BINANCE = "Binance"
    MACROTRENDS = "Macrotrends"
    EMPTY = "Empty" # use only for test

    # 4. ajouter votre source dans la liste
    SOURCES = [BINANCE, COINCODEX, GOOGLE_FINANCE, YAHOO, KAGGLE, MACROTRENDS, NASDAQ, EMPTY]

    # 5. ajouter votre nom de base de fichier pour la nouvelle source (la clé est la source)
    BASE_NAME = {
        BINANCE : "",
        COINCODEX : "coin_",
        GOOGLE_FINANCE : "GoogleFinance-",
        KAGGLE : "coin_",
        NASDAQ : "",
        MACROTRENDS : "MacroTrends_",
        YAHOO : "",
        EMPTY : ""
    }

    # Granularity Day, Hour
    GRANULARIY = ['', 'h']

    def __init__(self, dataDefaultPath='.', debug=False):
        self.data = {}
        self.debug = debug
        self.dataDefaultPath = dataDefaultPath

        if self.debug:
            print("init loader:", self.dataDefaultPath, self.debug)

        for source in Loader.SOURCES :
            self.data[source] = {}

    def loadFile(self, source, security, granularity='', dateStart=None, dateEnd=None):
        '''
        @param granulariy: default is day. Possible value is 'h'
        
        '''

        fileName = Loader.BASE_NAME[source]+security
        
        if(granularity != ''):
            fileName = fileName+'-'+granularity
        
        fileName = fileName+".csv"

        path = os.path.join(self.dataDefaultPath, "data", source, fileName)

        if self.debug:
            print("path =====", path, " & ", self.dataDefaultPath)
        
        if(os.path.isfile(path)):
            df = pd.read_csv(path, header=0)
            
            
            # df["Date"] = pd.to_datetime(df["Date"])
            # df["Date"] = df["Date"].dt.date
            # df= df.set_index("Date")

            #df.sort_index(axis=0, inplace=True)
            
            # print(source, " - ", df.columns)
            
            if source == Loader.BINANCE:
                # print("---------------", df.columns)
                df = df[['high', 'low', 'open', 'Close', 'Volume']]
                df.columns = ['High', 'Low', 'Open', 'Close', 'Volume']
                df['Market Cap'] = None
            elif source == Loader.COINCODEX:

                df["Date"] = pd.to_datetime(df["Start"])
                df["Date"] = df["Date"].dt.date
                df["Date"] = pd.to_datetime(df["Date"])
                df= df.set_index("Date")
                df.sort_index(axis=0, inplace=True)

                df = df[['High', 'Low', 'Open', 'Close', 'Volume', 'Market Cap']]
                df.columns = ['High', 'Low', 'Open', 'Close', 'Volume', 'Marketcap']
            elif source == Loader.KAGGLE:
                df = df[['High', 'Low', 'Open', 'Close', 'Volume', 'Marketcap']]
                df.columns = ['High', 'Low', 'Open', 'Close', 'Volume', 'Marketcap']
            elif source == Loader.NASDAQ:

                df["Date"] = pd.to_datetime(df["Date"])
                df["Date"] = df["Date"].dt.date
                df["Date"] = pd.to_datetime(df["Date"])
                df= df.set_index("Date")

                df.sort_index(axis=0, inplace=True)

                if 'Volume' not in df.columns:
                    df['Volume'] = 0

                # ['Close/Last', 'Volume', 'Open', 'High', 'Low']
                df = df[['High', 'Low', 'Open', 'Close/Last', 'Volume']]
                df.columns = ['High', 'Low', 'Open', 'Close', 'Volume']

                if(df['High'].dtypes == 'object'):
                    columnDollars = ['High', 'Low', 'Open', 'Close']
                    for col in columnDollars:
                        df[col] = df[col].str.replace('$', '')
                        df[col] = df[col].astype(float)

            elif source == Loader.YAHOO:
                # ['Open', 'High', 'Low', 'Close', 'Adj Close', 'Volume']
                
                df["Date"] = pd.to_datetime(df["Date"])
                df["Date"] = df["Date"].dt.date
                df["Date"] = pd.to_datetime(df["Date"])
                df= df.set_index("Date")

                df = df[['High', 'Low', 'Open', 'Close', 'Volume']]
                df.columns = ['High', 'Low', 'Open', 'Close', 'Volume']
            elif source == Loader.MACROTRENDS:
                df.rename(columns={"date": "Date"}, inplace=True)
                df["Date"] = pd.to_datetime(df["Date"])
                df["Date"] = df["Date"].dt.date
                df["Date"] = pd.to_datetime(df["Date"])
                df= df.set_index("Date")

                df = df[['high', 'low', 'open', 'close', 'volume']]
                df.columns = ['High', 'Low', 'Open', 'Close', 'Volume']
            elif source == Loader.GOOGLE_FINANCE:
                
                df["Date"] = pd.to_datetime(df["Date"], format='%d/%m/%Y %H:%M:%S')
                df["Date"] = df["Date"].dt.date
                df["Date"] = pd.to_datetime(df["Date"])
                df= df.set_index("Date")

                df.columns = ['High', 'Low', 'Open', 'Close', 'Volume']

            if(dateStart and dateEnd):
                datetimeStart = datetime.strptime(dateStart, '%Y-%m-%d').date()
                datetimeEnd = datetime.strptime(dateEnd, '%Y-%m-%d').date()
                
                df = df.loc[datetimeStart:datetimeEnd]
                

            self.data[source][security] = df
            
            if self.debug:
                print(source, "[", security, "]"," charged")

        else:
            if self.debug:
                print(source, "[", security, "]"," not charged")


    def loadSource(self, source, dateStart=None, dateEnd=None, forceEqualDate=False):
        
        # print("-----------------")
        
        for security in Loader.SECURITY:
                self.loadFile(source, security, dateStart=dateStart, dateEnd=dateEnd)

        # print("-----------------", ((dateStart and dateEnd) or forceEqualDate))
        
        numberOfSecurityLoaded = len(self.data[source].keys())
        
        if(numberOfSecurityLoaded>0 and ((dateStart and dateEnd) or forceEqualDate) ):
            startDates = []        
            endDates =  []

            for security, df in self.data[source].items():
                
                if df.shape[0] != 0:
                    startDates.append(df.index[0])
                    endDates.append(df.index[-1])
                else:
                    print("security ", security, " is empty")

            s = max(startDates)
            e = min(endDates)

            for security, df in self.data[source].items():
                self.data[source][security] = df.loc[s:e]
        
    def loadSources(self, dateStart=None, dateEnd=None, forceEqualDate=False):
        for source in Loader.SOURCES:
            self.loadSource(source, dateStart=dateStart, dateEnd=dateEnd, forceEqualDate=forceEqualDate)

    def isAvailable(self, source, nameSecurity):
        return source in self.data.keys() and nameSecurity in self.data[source].keys() 

    def get(self, source, nameSecurity, dateStart=None, dateEnd=None):
        if(source in self.data.keys() and nameSecurity in self.data[source].keys()):
            if(dateStart is None and dateEnd is None):
                return self.data[source][nameSecurity]
            else:
                df = self.data[source][nameSecurity]
                dateStart = datetime.strptime(dateStart, '%Y-%m-%d').date()
                dateEnd = datetime.strptime(dateEnd, '%Y-%m-%d').date()
                return df.loc[dateStart:dateEnd]
        
        self.loadFile(source, nameSecurity, dateStart=dateStart, dateEnd=dateEnd)
        return self.data[source][nameSecurity]

            

```
